```python
from scipy.stats import kruskal
import pandas as pd
from scipy.stats import wilcoxon
from itertools import combinations
import numpy as np
```

```python
# ambil F1-macro per fold dari skema final (Non-tuning)
f1_per_fold = {name: [f["f1"] for f in results_final[name]] for name in model_names}

stat, p_value = kruskal(*[f1_per_fold[name] for name in model_names])
kesimpulan = "Signifikan" if p_value < 0.05 else "Tidak Signifikan"

print("=" * 60)
print("UJI KRUSKAL-WALLIS ANTAR MODEL")
print("=" * 60)
print(f"Statistik : {stat:.4f}")
print(f"P-value   : {p_value:.4f}")
print(f"Kesimpulan: {kesimpulan}")

df_kw_model = pd.DataFrame([{
    "Statistik" : round(stat, 4),
    "P-value"   : round(p_value, 4),
    "Kesimpulan": kesimpulan
}])
print("\nTabel 4.3 - Hasil Uji Kruskal-Wallis")
print(df_kw_model.to_string(index=False))
```

```python
print("=" * 70)
print("UJI WILCOXON SIGNED-RANK ANTAR PASANGAN MODEL")
print("=" * 70)

wilcoxon_results = []

for model1, model2 in combinations(model_names, 2):
    f1_a = f1_per_fold[model1]
    f1_b = f1_per_fold[model2]

    try:
        stat, p_value = wilcoxon(f1_a, f1_b)
        keterangan = "Signifikan" if p_value < 0.05 else "Tidak Signifikan"
    except ValueError:
        stat, p_value, keterangan = np.nan, np.nan, "Tidak dapat dihitung (identik)"

    wilcoxon_results.append({
        "Pasangan Model": f"{model1} vs {model2}",
        "Statistik"     : stat,
        "P-value"       : round(p_value, 4) if not np.isnan(p_value) else np.nan,
        "Keterangan"    : keterangan,
    })

df_wilcoxon_model = pd.DataFrame(wilcoxon_results)
print("\nTabel 4.4 - Hasil Uji Wilcoxon Signed-Rank")
print(df_wilcoxon_model.to_string(index=False))
```
